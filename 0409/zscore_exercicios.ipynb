# Lista de Exercícios em Python — Z-Score

Observação: o desvio-padrão usado é o populacional (`ddof=0`), padrão do NumPy. No Pandas, `ddof=0` é informado explicitamente para manter o mesmo critério.

## Exercício 1 — Distância em passos de desvio-padrão

```python
media = 100
desvio = 5
valor = 115

distancia = valor - media
passos = distancia / desvio
z = (valor - media) / desvio

print(f"Distância entre o valor e a média: {distancia}")
print(f"Desvios-padrão que cabem nessa distância: {passos}")
print(f"Z-Score: {z}")
print(f"O valor {valor} está {abs(z):.1f} desvios-padrão {'acima' if z > 0 else 'abaixo'} da média {media}.")
```

## Exercício 2 — Acima ou abaixo da média?

```python
casos = [85, 100, 120]
media = 100
desvio = 10

for valor in casos:
    z = (valor - media) / desvio
    if z < 0:
        posicao = "abaixo da média"
    elif z == 0:
        posicao = "exatamente na média"
    else:
        posicao = "acima da média"
    print(f"Valor {valor} -> Z = {z:+.1f} -> {posicao}")

print()
print("Sinal positivo: o valor é maior que a média.")
print("Sinal negativo: o valor é menor que a média.")
print("Zero: o valor é igual à média.")
```

## Exercício 3 — Qual leitura é mais incomum?

```python
temperaturas = [21, 23, 25, 27, 29]
media = 25
desvio = 2

zs = []
for t in temperaturas:
    z = (t - media) / desvio
    zs.append(z)
    print(f"{t} °C -> Z = {z:+.1f}")

maior_abs = max(abs(z) for z in zs)
mais_distantes = [t for t, z in zip(temperaturas, zs) if abs(z) == maior_abs]

print()
print(f"Maior |Z|: {maior_abs}")
print(f"Temperaturas mais distantes da média: {mais_distantes}")
print("21 °C e 29 °C estão igualmente afastados (2 desvios-padrão), um abaixo e outro acima da média.")
```

## Exercício 4 — Latência de uma API

```python
import numpy as np

latencias = [98, 102, 101, 99, 100, 103, 97, 180]

media = np.mean(latencias)
desvio = np.std(latencias)
z_180 = (180 - media) / desvio

print(f"Média: {media:.2f} ms")
print(f"Desvio-padrão: {desvio:.2f} ms")
print(f"Z-Score de 180 ms: {z_180:.2f}")

if abs(z_180) > 3:
    print("Pela regra |Z| > 3, a latência de 180 ms merece investigação.")
else:
    print("Pela regra |Z| > 3, a latência de 180 ms não é sinalizada.")
    print("Mesmo sendo visivelmente diferente, o próprio valor 180 aumenta a média e o desvio-padrão,")
    print("o que reduz o seu Z-Score. Em conjuntos pequenos, isso pode mascarar um valor atípico.")

print()
print("Lembrete: investigar não significa apagar automaticamente. O valor precisa ser analisado antes de qualquer decisão.")
```

## Exercício 5 — Monitoramento de CPU com classificação

```python
import numpy as np

cpu = [42, 45, 47, 44, 46, 43, 48, 92]

media = np.mean(cpu)
desvio = np.std(cpu)

print(f"Média: {media:.2f}%")
print(f"Desvio-padrão: {desvio:.2f}%")
print()

for leitura in cpu:
    z = (leitura - media) / desvio
    classificacao = "Investigar" if abs(z) > 3 else "Comum"
    print(f"{leitura} -> {z:+.2f} -> {classificacao}")

print()
print("A leitura de 92% se destaca, mas não passa de |Z| = 3: com apenas 8 leituras,")
print("o próprio pico infla o desvio-padrão e limita o Z-Score máximo possível.")
```

## Exercício 6 — O mesmo valor em dois contextos

```python
media_a = 100
desvio_a = 2
media_b = 100
desvio_b = 20
valor = 110

z_a = (valor - media_a) / desvio_a
z_b = (valor - media_b) / desvio_b

print(f"Z-Score no Grupo A: {z_a}")
print(f"Z-Score no Grupo B: {z_b}")
print()
print(f"No Grupo A, 110 está {z_a} desvios-padrão acima da média: muito incomum.")
print(f"No Grupo B, 110 está {z_b} desvio-padrão acima da média: comum.")
print()
print("A distância absoluta é a mesma (10 unidades), mas o Z-Score mede essa distância")
print("em relação à dispersão de cada grupo. No Grupo A os dados variam pouco, então 10 unidades")
print("é um afastamento enorme; no Grupo B os dados variam muito, então 10 unidades é pouco.")
```

## Exercício 7 — Função de interpretação

```python
def interpretar_z(z):
    if abs(z) > 3:
        return "Investigar"
    if z < 0:
        return "Abaixo da média"
    if z > 0:
        return "Acima da média"
    return "Na média"

valores_z = [-3.5, -1.2, 0, 0.8, 3.7]

for z in valores_z:
    print(f"Z = {z:+} -> {interpretar_z(z)}")
```

## Exercício 8 — Z-Score em um DataFrame

```python
import pandas as pd
import numpy as np

dados = {
    "Usuario": ["ana", "bruno", "carla", "diego", "eva", "fabio"],
    "Requisicoes": [120, 135, 128, 122, 130, 400]
}
df = pd.DataFrame(dados)

media = df["Requisicoes"].mean()
desvio = df["Requisicoes"].std(ddof=0)

print(f"Média: {media:.2f}")
print(f"Desvio-padrão: {desvio:.2f}")

df["Z_Score"] = ((df["Requisicoes"] - media) / desvio).round(2)
df["Status"] = np.where(df["Z_Score"].abs() > 3, "Investigar", "Comum")

df
```

```python
investigar = df[df["Status"] == "Investigar"]

if investigar.empty:
    print("Nenhuma linha ultrapassou |Z| > 3.")
    print(f"O usuário fabio tem o maior Z-Score ({df['Z_Score'].max()}), bem acima dos demais,")
    print(f"mas com apenas {len(df)} registros o Z-Score máximo possível é raiz({len(df)} - 1) = {np.sqrt(len(df) - 1):.2f}.")
    print("Aqui a regra |Z| > 3 não consegue sinalizar nada; vale considerar outra técnica, como o IQR.")
else:
    display(investigar)
```

## Exercício 9 — Comparação entre IQR e Z-Score

```python
import numpy as np

dados = [10, 11, 12, 12, 13, 13, 14, 15, 30]

q1 = np.percentile(dados, 25)
q3 = np.percentile(dados, 75)
iqr = q3 - q1
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

media = np.mean(dados)
desvio = np.std(dados)
z_30 = (30 - media) / desvio

print(f"Q1: {q1}")
print(f"Q3: {q3}")
print(f"IQR: {iqr}")
print(f"Limite inferior: {limite_inferior}")
print(f"Limite superior: {limite_superior}")
print()
print(f"Média: {media:.2f}")
print(f"Desvio-padrão: {desvio:.2f}")
print(f"Z-Score de 30: {z_30:.2f}")
print()
print(f"IQR: 30 {'está fora' if 30 > limite_superior else 'está dentro'} dos limites -> {'possível outlier' if 30 > limite_superior else 'comum'}.")
print(f"Z-Score: |Z| = {abs(z_30):.2f} {'>' if abs(z_30) > 3 else '<='} 3 -> {'investigar' if abs(z_30) > 3 else 'não sinalizado'}.")
```

```python
print("Conclusão:")
print("O IQR usa quartis, que quase não são afetados pelo valor 30, e por isso o identifica como atípico.")
print("O Z-Score usa média e desvio-padrão, que são puxados pelo próprio 30, e por isso o sinal fica mais fraco.")
print("Técnicas diferentes analisam o mesmo dado por critérios diferentes e podem chegar a conclusões diferentes.")
```

## Exercício 10 — Mini análise de eventos de segurança

```python
import pandas as pd

dados = {
    "Evento": ["A", "B", "C", "D", "E", "F", "G"],
    "Tentativas_Login": [3, 4, 2, 5, 3, 4, 40]
}
df = pd.DataFrame(dados)

media = df["Tentativas_Login"].mean()
desvio = df["Tentativas_Login"].std(ddof=0)
df["Z_Score"] = ((df["Tentativas_Login"] - media) / desvio).round(2)

print(f"Média: {media:.2f}")
print(f"Desvio-padrão: {desvio:.2f}")

df
```

```python
sinalizados = df[df["Z_Score"].abs() > 3]

if sinalizados.empty:
    maior = df.loc[df["Z_Score"].abs().idxmax()]
    print("Nenhum evento ultrapassou |Z| > 3.")
    print(f"O evento {maior['Evento']} tem o maior Z-Score ({maior['Z_Score']}) e está muito acima dos demais;")
    print("com apenas 7 eventos, o próprio valor extremo infla o desvio-padrão e limita o Z-Score.")
else:
    display(sinalizados)

print()
print("Em segurança, um evento incomum pode ser justamente o dado mais importante da análise:")
print("40 tentativas de login podem indicar um ataque de força bruta ou uso indevido de credenciais.")
print("Aqui o objetivo não é limpar o dado para melhorar a estatística, e sim encontrar a anomalia.")
print("Remover esse valor seria apagar a evidência do problema que se quer detectar.")
```
