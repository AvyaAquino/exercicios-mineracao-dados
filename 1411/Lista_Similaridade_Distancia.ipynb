# Lista de Exercícios — Similaridade e Distância em Python

**Fórmula da distância Euclidiana** entre dois pontos $p$ e $q$ com $n$ atributos:

$$d(p, q) = \sqrt{(p_1 - q_1)^2 + (p_2 - q_2)^2 + \dots + (p_n - q_n)^2}$$

Quanto **menor** a distância, **mais parecidos** são os registros.

# Parte A — Sem uso de funções prontas de distância

## Exercício 1 — Distância Euclidiana entre dois clientes

```python
cliente_a = [25, 2]
cliente_b = [28, 6]

dif_idade = cliente_a[0] - cliente_b[0]

dif_compras = cliente_a[1] - cliente_b[1]

quad_idade = dif_idade ** 2
quad_compras = dif_compras ** 2

soma = quad_idade + quad_compras

distancia = soma ** 0.5

print(f"Diferença de idade:   {dif_idade}  -> ao quadrado: {quad_idade}")
print(f"Diferença de compras: {dif_compras} -> ao quadrado: {quad_compras}")
print(f"Soma dos quadrados:   {soma}")
print(f"Distância Euclidiana entre A e B: {distancia:.2f}")
```

## Exercício 2 — Qual cliente é mais semelhante ao cliente A?

```python
cliente_a = [25, 2]
cliente_b = [27, 3]
cliente_c = [45, 10]
cliente_d = [52, 12]

dist_ab = ((cliente_a[0] - cliente_b[0]) ** 2 + (cliente_a[1] - cliente_b[1]) ** 2) ** 0.5

dist_ac = ((cliente_a[0] - cliente_c[0]) ** 2 + (cliente_a[1] - cliente_c[1]) ** 2) ** 0.5

dist_ad = ((cliente_a[0] - cliente_d[0]) ** 2 + (cliente_a[1] - cliente_d[1]) ** 2) ** 0.5

distancias = {"B": dist_ab, "C": dist_ac, "D": dist_ad}

mais_proximo = "B"
for cliente, d in distancias.items():
    if d < distancias[mais_proximo]:
        mais_proximo = cliente

for cliente, d in distancias.items():
    print(f"Distância A-{cliente}: {d:.2f}")
print(f"\nCliente mais próximo de A: {mais_proximo} ({distancias[mais_proximo]:.2f})")

# POR QUE A MENOR DISTÂNCIA INDICA MAIOR SEMELHANÇA?
# A distância Euclidiana mede o "tamanho da diferença" entre os atributos dos dois
# clientes (idade e compras por mês). Se a distância é pequena, os valores de idade
# e de compras são parecidos, ou seja, os perfis quase coincidem. Se a distância é
# grande, os clientes diferem muito em pelo menos um atributo. Por isso, quanto
# menor a distância, maior a semelhança -- considerando APENAS esses dois atributos.
```

## Exercício 3 — Distância Euclidiana com três atributos

```python
aluno_a = [8.0, 92, 10]
aluno_b = [7.8, 90, 9]
aluno_c = [5.2, 68, 3]

dist_ab = ((aluno_a[0] - aluno_b[0]) ** 2 +
           (aluno_a[1] - aluno_b[1]) ** 2 +
           (aluno_a[2] - aluno_b[2]) ** 2) ** 0.5

dist_ac = ((aluno_a[0] - aluno_c[0]) ** 2 +
           (aluno_a[1] - aluno_c[1]) ** 2 +
           (aluno_a[2] - aluno_c[2]) ** 2) ** 0.5

print(f"Distância A-B: {dist_ab:.2f}")
print(f"Distância A-C: {dist_ac:.2f}")

if dist_ab < dist_ac:
    print("O aluno B está mais próximo de A.")
else:
    print("O aluno C está mais próximo de A.")
```

# Parte B — Com uso de funções prontas

## Exercício 4 — Comparando vários clientes com `math.dist()`

```python
import math

clientes = {
    "A": [25, 2],
    "B": [27, 3],
    "C": [30, 5],
    "D": [45, 10],
    "E": [52, 12]
}

distancias = {}
for nome, dados in clientes.items():
    if nome == "A":
        continue
    distancias[nome] = math.dist(clientes["A"], dados)
    print(f"Distância A-{nome}: {distancias[nome]:.2f}")

mais_proximo = min(distancias, key=distancias.get)

print(f"\nO cliente {mais_proximo} possui o perfil mais próximo de A, "
      f"considerando idade e compras por mês (distância = {distancias[mais_proximo]:.2f}).")
```

## Exercício 5 — Influência da escala e normalização

```python
import math
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
    "Cliente": ["A", "B", "C", "D", "E"],
    "Idade":   [20, 25, 40, 55, 70],
    "Renda":   [2000, 3000, 6000, 9000, 12000]
}).set_index("Cliente")
df
```

```python
dist_original = math.dist(df.loc["A"], df.loc["B"])
print(f"Distância A-B (original): {dist_original:.2f}")
```

```python
scaler = MinMaxScaler()
df_norm = pd.DataFrame(
    scaler.fit_transform(df[["Idade", "Renda"]]),
    columns=["Idade", "Renda"],
    index=df.index
)

df_norm.round(3)
```

```python
dist_norm = math.dist(df_norm.loc["A"], df_norm.loc["B"])
print(f"Distância A-B (normalizada): {dist_norm:.4f}")

print(f"\nOriginal:    {dist_original:.2f}")
print(f"Normalizada: {dist_norm:.4f}")

dif_idade = (df.loc["A", "Idade"] - df.loc["B", "Idade"]) ** 2
dif_renda = (df.loc["A", "Renda"] - df.loc["B", "Renda"]) ** 2
print(f"\nContribuição (dif²) da idade: {dif_idade}")
print(f"Contribuição (dif²) da renda: {dif_renda}")

# POR QUE A RENDA DOMINA COM OS VALORES ORIGINAIS?
# A renda está na casa dos milhares (2.000 a 12.000), enquanto a idade está na casa
# das dezenas (20 a 70). Entre A e B a diferença de renda é 1000 e a de idade é 5.
# Ao elevar ao quadrado: 1000² = 1.000.000 contra 5² = 25. A idade praticamente
# "some" do cálculo -- a distância (≈1000,01) é quase igual à diferença de renda.
# Ou seja, a renda pesa muito mais só porque usa números maiores, não porque seja
# mais importante. Com a normalização Min-Max, os dois atributos passam a variar
# de 0 a 1 e contribuem de forma equilibrada para a distância.
```

## Exercício 6 — Matriz de distâncias entre registros

```python
import math
import pandas as pd

registros = {
    "A": [1, 1],
    "B": [4, 5],
    "C": [2, 2],
    "D": [5, 4]
}

nomes = list(registros.keys())
matriz = []
for linha in nomes:
    valores_linha = []
    for coluna in nomes:
        valores_linha.append(math.dist(registros[linha], registros[coluna]))
    matriz.append(valores_linha)

df_dist = pd.DataFrame(matriz, index=nomes, columns=nomes)
df_dist.round(2)
```

```python
menor = None
for i in range(len(nomes)):
    for j in range(i + 1, len(nomes)):
        d = df_dist.iloc[i, j]
        if menor is None or d < menor:
            menor = d

pares = []
for i in range(len(nomes)):
    for j in range(i + 1, len(nomes)):
        if math.isclose(df_dist.iloc[i, j], menor):
            pares.append(f"{nomes[i]}-{nomes[j]}")

print(f"Menor distância diferente de zero: {menor:.2f}")
print(f"Par(es) de registros mais próximo(s): {', '.join(pares)}")
#teve um empate
```

```python
diagonal = [float(df_dist.iloc[i, i]) for i in range(len(nomes))]
print("Diagonal:", diagonal)
print("Diagonal só tem zeros?", all(v == 0 for v in diagonal))

print("Matriz é simétrica?", df_dist.equals(df_dist.T))

# POR QUE A DIAGONAL É ZERO?
# A diagonal compara cada registro com ele mesmo: d(A,A). Todas as diferenças entre
# os atributos são 0, então a soma dos quadrados é 0 e a raiz também é 0.
# Um registro é sempre idêntico a si mesmo.
#
# POR QUE d(A,B) = d(B,A)?
# Na fórmula, cada diferença é elevada ao quadrado: (a - b)² = (b - a)².
# Trocar a ordem só inverte o sinal da diferença, e o quadrado elimina o sinal.
# Por isso a distância de A até B é igual à de B até A, e a matriz é simétrica
# (o triângulo de cima é o espelho do triângulo de baixo).
```
